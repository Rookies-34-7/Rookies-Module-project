# Lifestyle Risk Index, Sleep Risk Index 상관관계 분석
# Lifestyle Risk Index

```python
# 임포트
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#데이터 불러오기
data = pd.read_csv('../risk_data/Sleep Health and Lifestyle Dataset.csv')
data.head()
```

```python
#상관 계수 계산, heatmap 출력
correlation = data.corr(numeric_only=True)
correlation
```

```python
lifestyle_corr = correlation['Lifestyle_Risk_Index']
lifestyle_corr = lifestyle_corr.sort_values(ascending=False)
lifestyle_corr
```

```python
# 자기자신 제거
lifestyle_corr = lifestyle_corr.drop('Lifestyle_Risk_Index')
lifestyle_corr
```

```python
#절댓값 계산
abs_corr = lifestyle_corr.abs()
abs_corr = abs_corr.sort_values(ascending=False)
abs_corr
```

```python
#top 10 상관계수 절댓값 높은 거 
top10 = abs_corr.head(10)
top10
top10.plot(kind='bar')
plt.title('Correlation with Lifestyle Risk Index')
plt.ylabel('Absolute Correlation')
plt.show()
```

```python
#선택하려는 데이터들
selected_numeric = [
    'Age',
    'BMI',
    'Sleep_Duration',
    'Screen_Time_Hours',
    'Caffeine_Intake_mg',
    'Physical_Activity_Minutes',
    'Lifestyle_Risk_Index'
]
selected_data = data[selected_numeric]
selected_corr = selected_data.corr()
selected_corr
```

```python
#정렬
selected_corr['Lifestyle_Risk_Index'].sort_values(ascending=False)
```

```python
#스크린 타임 산점도
plt.scatter(data['Screen_Time_Hours'], data['Lifestyle_Risk_Index'])
plt.xlabel('Screen Time Hours')
plt.ylabel('Lifestyle Risk Index')
plt.title('Screen Time vs Lifestyle Risk')
plt.show()
```

```python
# 카페인 섭취 산점도
plt.scatter(data['Caffeine_Intake_mg'], data['Lifestyle_Risk_Index'])
plt.xlabel('Caffeine Intake')
plt.ylabel('Lifestyle Risk Index')
plt.title('Caffeine vs Lifestyle Risk')
plt.show()
```

```python
#운동량 산점도
plt.scatter(data['Physical_Activity_Minutes'], data['Lifestyle_Risk_Index'])
plt.xlabel('Physical Activity Minutes')
plt.ylabel('Lifestyle Risk Index')
plt.title('Physical Activity vs Lifestyle Risk')
plt.show()
```

```python
#남녀 데이터 차이
data.groupby('Gender')['Lifestyle_Risk_Index'].mean()
```

```python
#흡연 데이터 차이
data.groupby('Smoking_Status')['Lifestyle_Risk_Index'].mean()
```

```python
#24시간 내 알콜 데이터 차이 
data.groupby('Alcohol_Consumption')['Lifestyle_Risk_Index'].mean()
```

```python
stress_map = {4: 'Low', 7: 'Medium', 9: 'High'}
data['Stress_Category'] = data['Stress_Level'].map(stress_map)
stress_mean = data.groupby('Stress_Category')['Lifestyle_Risk_Index'].mean()
stress_mean
```

```python
model_numeric = [
    'Age',
    'BMI',
    'Sleep_Duration',
    'Screen_Time_Hours',
    'Caffeine_Intake_mg',
    'Physical_Activity_Minutes'
]
model_corr = data[model_numeric + ['Lifestyle_Risk_Index']].corr()
model_corr['Lifestyle_Risk_Index'].sort_values(ascending=False)
```

#Sleep_Risk_Index

```python
#상관 계수 계산
sleep_corr = data.corr(numeric_only=True)
sleep_corr
```

```python
#자기자신 제거
sleep_risk_corr = sleep_corr["Sleep_Risk_Index"]
sleep_risk_corr = sleep_risk_corr.drop("Sleep_Risk_Index")
```

```python
#abs로 정렬
sleep_risk_corr = sleep_risk_corr.reindex(
    sleep_risk_corr.abs().sort_values(ascending=False).index
)
sleep_risk_corr
```

```python
#상위 데이터 
sleep_features = [
    "Stress_Level",
    "Sleep_Quality",
    "Daytime_Sleepiness",
    "Sleep_Efficiency",
    "Workload_Score",
    "Anxiety_Level",
    "Sleep_Latency",
    "Sleep_Duration",
    "Depression_Score"
]
sleep_data = data[sleep_features + ["Sleep_Risk_Index"]]
```

```python
#상관계수 확인
sleep_data.corr()
```

```python
# 자기 자신 제거
#sleep_feature_corr = sleep_feature_corr.drop("Sleep_Risk_Index")

# 절댓값 기준으로 상관관계가 강한 순서대로 정렬
sleep_feature_corr = sleep_feature_corr.reindex(
    sleep_feature_corr.abs().sort_values(ascending=False).index
)
sleep_feature_corr
```

```python
feature_corr = data[sleep_features].corr()
feature_corr
```
