```python
import pandas as pd

# 데이터 불러오기
df = pd.read_csv("Sleep_health_and_lifestyle_dataset.csv")

# 앞부분 확인
df.head()
```

```python
# 데이터 크기 확인
df.shape
```

```python
# 데이터 정보 확인
df.info()
```

```python
# 결측치 확인
df.isnull().sum()
```

```python
# 중복 데이터 확인
df.duplicated().sum()
```

```python
# 수면장애 종류와 개수 확인
df["Sleep Disorder"].value_counts(dropna=False)
```

```python
# 수면장애 결측치는 'None'으로 변경
df["Sleep Disorder"] = df["Sleep Disorder"].fillna("None")
```

```python
df["Sleep Disorder"].value_counts()
```

```python
# Person ID 제거
df = df.drop("Person ID", axis=1)

# 컬럼 확인
df.columns
```

```python
# BMI Category 종류 확인
df["BMI Category"].value_counts()
```

```python
# BMI Category 값 통일
df["BMI Category"] = df["BMI Category"].replace("Normal Weight", "Normal")

# 변경 결과 확인
df["BMI Category"].value_counts()
```

```python
# Blood Pressure 값 확인
df["Blood Pressure"].head(10)
```

```python
# Blood Pressure를 수축기/이완기 혈압으로 분리
df[["Systolic BP", "Diastolic BP"]] = df["Blood Pressure"].str.split("/", expand=True)

# 숫자형으로 변환
df["Systolic BP"] = df["Systolic BP"].astype(int)
df["Diastolic BP"] = df["Diastolic BP"].astype(int)

# 결과 확인
df[["Blood Pressure", "Systolic BP", "Diastolic BP"]].head()
```

```python
# 기존 Blood Pressure 컬럼 제거
df = df.drop("Blood Pressure", axis=1)

# 현재 컬럼 확인
df.columns
```

```python
# 문자형 컬럼 확인
df.select_dtypes(include="object").columns
```

```python
# 문자형 컬럼 값 확인
print("Gender")
print(df["Gender"].value_counts())

print("\nOccupation")
print(df["Occupation"].value_counts())

print("\nBMI Category")
print(df["BMI Category"].value_counts())

print("\nSleep Disorder")
print(df["Sleep Disorder"].value_counts())
```
